```python
import kagglehub
path = kagglehub.dataset_download("skaspari/makemore-names")
```

```python
words = open(f'{path}/names.txt','r').read().splitlines()
words
```

```python
# Converting chars into numbers bcz pytorch workss with tensors
chars = sorted(set("".join(words)))

# String to integer
stoi = {s:i+1 for i,s in enumerate(chars)}
stoi["."] = 0

# Integer to string
itos = {i:s for s, i in stoi.items()}

vocab_size = len(stoi)
```

```python
import torch
N = torch.zeros((vocab_size,vocab_size),dtype=torch.int32) #27x27 matrix of a character mapping
N
```

```python
print("Building a  simple bigram")
for word in words:
  chars = ["."] + list(word) + ["."]
  for ch1,ch2 in zip(chars,chars[1:]):
    ix1 = stoi[ch1]
    ix2  = stoi[ch2]

    N[ix1,ix2] += 1


N.shape
```

```python
g = torch.Generator().manual_seed(2147483647)
for _ in range(20):
  ix = 0
  output = []

  while True:
      # counts  of possible next characters
      counts = N[ix].float() #Entire row corressponding to N[ix]
      # Convert counts -> probabilities e.g a=100,b=50 whatever..
      probabilities = counts/sum(counts) #It's a prob dist. now..

      # Sample next character (Pick one char acc to the probs in p) /Random sampling..
      ix = torch.multinomial(
          probabilities,
          num_samples=1,
          replacement=True,
          generator=g
      ).item()

      if ix==0: #. means end of name..
        break

      output.append(itos[ix]) #append the




  print("".join(output))
```

```python
# Neural Bigram Model
import torch
import torch.nn.functional as F

# Create training examples
xs = []
ys = []

for word in words:
  chars = ["."] + list(word) + ["."]

  for ch1,ch2 in zip(chars,chars[1:]):
    xs.append(stoi[ch1]) #current character
    ys.append(stoi[ch2]) #next character


xs = torch.tensor(xs)
ys  = torch.tensor(ys)

print(xs.shape)
print(ys.shape)
```

```python
# Initializign the weights/model..

g = torch.Generator().manual_seed(2147483647)

# This iss where we will store weights, random values nxn
W = torch.randn(
    (27,27),
    generator=g,
    requires_grad=True
)
```

```python
for step in range(100):
   # Forward pass
   xenc = F.one_hot(xs,num_classes=27).float() #input features , xenc = (number of examples,27) ..W=(27,27)
   logits  = xenc@W #(logits=(number_of_examples,27))
  #  we get [2.0,1.0,0.0]..thesse are prob

  #  Softmax (turn into valid prob bw 0 and 1), which sum to 1
   counts = logits.exp()  #exp does-> large number = much larger +ve number, smaller sscore-> smaller number..
   probs = counts/counts.sum(dim=1,keepdim=True)

  #Negative log likelihood

  # High prob for correct answer -> small NLL. ..
  # lOW prob for correct answer -> large NLL...
   correct_probs = probs[torch.arange(len(xs)),ys] #How good was the probability the model assigned to the correct answer?
  #  e.g p(m|a) = 0.8, then -log(0.8) = 0.223 small loss, but if 0.1 it gives 2.303

    # we want mean -ve log likelihood, and that's what we're using ass our training loss
   nll = -correct_probs.log().mean()  #we want a +ve loss

  #  L2 Regularization
   reg = 0.01*(W**2).mean()

   loss = nll+reg

  #  Backward pass
   W.grad = None
   loss.backward()

    # Gradient descent/loss update
   W.data += -0.1*W.grad
   print(f"Step : {step}: loss={loss.item():.4f}")
```

```python
# GENERATING NAMES:
for _ in range(10):
  ix = 0
  output  = []

  while True:
    # Current character -> one hot
    x = F.one_hot(torch.tensor([ix]),num_classes=27).float()

    # Neural network
    logits = x@W

    # Softmax
    probs = F.softmax(logits,dim=1) #Softmax is the function that takes the model's raw scores (logits) and turns them into probabilities that add up to 1...[2.0, 1.0, 0.0]# ↓ softmax # [0.665, 0.245, 0.090]

    # Sample next char
    ix = torch.multinomial(probs,num_samples=1,replacement=True,generator=g).item()

    # End
    if ix==0:
      break

    output.append(itos[ix])

  print("".join(output))
```
